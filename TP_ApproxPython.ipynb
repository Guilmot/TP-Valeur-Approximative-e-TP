```python
# on importe le module pyplot du package

import matplotlib.pyplot as plt
import numpy as np # matplotlib s'appuie sur numpy pour fonctionner
# numpy permet aussi de disposer de fonctionnalités mathématiques

## fonction f
def f(x):
    return np.log(x) - 1

## configurations de la courbe de f
plt.figure()
x=np.linspace(1,6,num=12) ## definie une valeur x qui est un numpy linspace dans l'intervalle 1,6 avec un pas de 0.5 car 0.5*12 = 6
plt.plot(x,f(x),"r") ## on trace la courbe en prenant la valeur retourné par la fonction f(x)
plt.show()
```

1. Calculer numériquement les valeurs de ai et bi pour i = 0,1,2 .

```python
## une fonction qui permet d'afficher le résultat des an, bn et cn des trois listes
def afficheListe(listeA,listeB,listeC,roundDecimal):
    for i in range (len(listeA)):
        print(f"a{i} = {round(listeA[i],roundDecimal)}; b{i} = {round(listeB[i],roundDecimal)}; c{i} = {round(listeC[i],roundDecimal)}")

## une fonction qui va calculez tous les a,b et c jusqu'a n
def rechercheDichotomie(n,a0,b0):
    ## on definie les listes contenant les valeurs an,bn et cn
    liste_a = []
    liste_b = []
    liste_c = []

    ## c0 est calculé
    c0 = (a0 + b0)/2
    
    ## on ajoute les valeurs de a0,b0 et c0 initial dans les listes correspondantes
    liste_a.append(a0)
    liste_b.append(b0)
    liste_c.append(c0)
    for i in range(0,n):
        ## cn+1 est toujours égal à (an + bn)/2
        c_plus1 = (liste_a[i] + liste_b[i])/2

        ## an+1 est égal à an si f(an) x f(cn+1) <= 0 sinon an+1 est égal à cn+1
        a_plus1 = None
        if f(liste_a[i]) * f(c_plus1) <= 0:
            a_plus1 = liste_a[i]
        else:
            a_plus1 = c_plus1

        ## bn+1 est égal à cn+1 si f(an) * f(cn+1) <= 0 sinon bn+1 est égal à bn
        b_plus1 = None
        if f(liste_a[i]) * f(c_plus1) <= 0:
            b_plus1 = c_plus1
        else:
            b_plus1 = liste_a[i]

        ## on ajoute dans les listes correspondantes les nouvelles valeurs de an+1, bn+1 et cn+1
        liste_a.append(a_plus1)
        liste_b.append(b_plus1)
        liste_c.append(c_plus1)

    return liste_a,liste_b,liste_c        
```

```python
## Pour la question 2.1 on veut savoir les ai et bi pour i = 0,1,2

## on recupere la liste de an, bn et cn retourné par la fonction rechercheDichotomie
listeA,listeB,listeC = rechercheDichotomie(3,1,6)

## on affiche le résultat
afficheListe(listeA,listeB,listeC,3)
```
